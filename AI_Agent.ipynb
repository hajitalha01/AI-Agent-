```python
# Simple AI Agent Example

class MathAgent:
    def solve(self, question):
        try:
            # User ka expression evaluate karega
            answer = eval(question)
            return f"Answer: {answer}"
        except:
            return "Sorry, main is question ko solve nahi kar saka."

# Agent create karo
agent = MathAgent()

# User se input lo
question = input("Math question likho (e.g. 5+3*2): ")

# Agent se answer lo
result = agent.solve(question)

print(result)
```
