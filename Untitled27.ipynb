```python
import numpy as np
import random
import torch
import torch.nn as nn
import torch.optim as optim
from collections import deque

# --- CONFIGURATIONS ---
MAZE_ROWS = 13
MAZE_COLS = 13
STATE_SIZE = 4   # UP, DOWN, LEFT, RIGHT ki walls/paths (0 for empty, 1 for wall)
ACTION_SIZE = 4  # 0=UP, 1=DOWN, 2=LEFT, 3=RIGHT

# =====================================================================
# 1. ENVIRONMENT CLASS
# =====================================================================
class MazeEnvironment:
    def __init__(self):
        self.start_pos = (0, 11)
        self.target_pos = (12, 6)
        self.reset()

    def generate_random_maze(self):
        # 0 = Khali rasta, 1 = Deewar
        maze = np.zeros((MAZE_ROWS, MAZE_COLS), dtype=int)

        # Randomly deewarein khari karna (30% density)
        for r in range(MAZE_ROWS):
            for c in range(MAZE_COLS):
                if random.random() < 0.3:
                    maze[r, c] = 1

        # Start aur Target ko hamesha khali (0) rakhna hai
        maze[self.start_pos[0], self.start_pos[1]] = 0
        maze[self.target_pos[0], self.target_pos[1]] = 0
        return maze

    def reset(self):
        self.maze = self.generate_random_maze()
        self.agent_pos = list(self.start_pos)
        self.score = 0
        self.punish_count = 0
        self.steps = 0  # To prevent infinite loops
        return self.get_state()

    def get_state(self):
        r, c = self.agent_pos
        # Check binary status: 1 agar deewar ya boundary hai, 0 agar khali rasta hai
        up = 1 if r == 0 or self.maze[r-1, c] == 1 else 0
        down = 1 if r == MAZE_ROWS-1 or self.maze[r+1, c] == 1 else 0
        left = 1 if c == 0 or self.maze[r, c-1] == 1 else 0
        right = 1 if c == MAZE_COLS-1 or self.maze[r, c+1] == 1 else 0

        return np.array([up, down, left, right], dtype=np.float32)

    def step(self, action):
        self.steps += 1
        r, c = self.agent_pos
        new_r, new_c = r, c

        # Actions mapping
        if action == 0: new_r -= 1    # UP
        elif action == 1: new_r += 1  # DOWN
        elif action == 2: new_c -= 1  # LEFT
        elif action == 3: new_c += 1  # RIGHT

        hit_wall = False
        # Check boundary or wall hit
        if new_r < 0 or new_r >= MAZE_ROWS or new_c < 0 or new_c >= MAZE_COLS or self.maze[new_r, new_c] == 1:
            hit_wall = True
        else:
            self.agent_pos = [new_r, new_c]

        # Rules & Reward Logic
        done = False
        if hit_wall:
            reward = -1
            self.score -= 1
        elif tuple(self.agent_pos) == self.target_pos:
            reward = 100  # Target tak pohonchne ka bara inaam
            self.score += 100
            done = True
        else:
            reward = 1  # Khali rasta mila
            self.score += 1

        # Punish rule: Agar score 50 touch ho jaye
        if self.score >= 50 and not done:
            self.score -= 25
            self.punish_count += 1
            reward = -25

        # Infinite loop se bachne ke liye limit
        if self.steps >= 200:
            done = True

        next_state = self.get_state()
        return next_state, reward, done

# =====================================================================
# 2. DQN CLASS (Neural Network)
# =====================================================================
class DQN_Network(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(DQN_Network, self).__init__()
        self.fc1 = nn.Linear(input_dim, 64)
        self.fc2 = nn.Linear(64, 64)
        self.fc3 = nn.Linear(64, output_dim)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        return self.fc3(x)

# =====================================================================
# 3. AGENT CLASS
# =====================================================================
class DQNAgent:
    def __init__(self):
        self.memory = deque(maxlen=2000) # Replay Buffer
        self.gamma = 0.95                # Discount factor
        self.epsilon = 1.0               # Exploration rate
        self.epsilon_min = 0.01
        self.epsilon_decay = 0.995       # Ab yeh har episode par scale hoga properly
        self.batch_size = 32

        # Network aur Optimizer
        self.model = DQN_Network(STATE_SIZE, ACTION_SIZE)
        self.optimizer = optim.Adam(self.model.parameters(), lr=0.001)
        self.criterion = nn.MSELoss()

    def remember(self, state, action, reward, next_state, done):
        self.memory.append((state, action, reward, next_state, done))

    def act(self, state):
        # Exploration vs Exploitation
        if np.random.rand() <= self.epsilon:
            return random.randrange(ACTION_SIZE) # Random action

        state = torch.FloatTensor(state).unsqueeze(0)
        with torch.no_grad():
            q_values = self.model(state)
        return torch.argmax(q_values).item() # Best action from Neural Network

    def replay(self):
        if len(self.memory) < self.batch_size:
            return

        # Memory se random batch uthana
        minibatch = random.sample(self.memory, self.batch_size)

        for state, action, reward, next_state, done in minibatch:
            state_t = torch.FloatTensor(state)
            next_state_t = torch.FloatTensor(next_state)

            # Target Q-value calculate karna
            target = reward
            if not done:
                # Bellman Equation
                target = reward + self.gamma * torch.max(self.model(next_state_t)).item()

            # Current Q-values nikalna
            current_q = self.model(state_t)
            target_f = current_q.clone().detach()
            target_f[action] = target

            # Neural Network ko train (Backpropagation) karna
            self.optimizer.zero_grad()
            loss = self.criterion(current_q, target_f)
            loss.backward()
            self.optimizer.step()

    def save_model(self, name):
        torch.save(self.model.state_dict(), name)

# =====================================================================
# 4. EXECUTION LOOP (Training)
# =====================================================================
if __name__ == "__main__":
    env = MazeEnvironment()
    agent = DQNAgent()
    episodes = 500

    print("AI Training Shuru Ho Rahi Hai...\n")

    for episode in range(1, episodes + 1):
        state = env.reset()
        done = False

        while not done:
            action = agent.act(state)
            next_state, reward, done = env.step(action)
            agent.remember(state, action, reward, next_state, done)

            # Har step par replay memory se neural net weights optimize honge
            agent.replay()

            state = next_state

        # [FIXED] Epsilon decay ko har episode ke baaher shift kar diya gaya hai
        if agent.epsilon > agent.epsilon_min:
            agent.epsilon *= agent.epsilon_decay

        # Progress tracking
        if episode % 10 == 0:
            print(f"Episode {episode}/{episodes} | Final Score: {env.score} | Punishes: {env.punish_count} | Epsilon: {agent.epsilon:.4f}")

    # Model save karna
    agent.save_model("smart_agent.pth")
    print("\nTraining Khatam! Model 'smart_agent.pth' ke naam se save ho gaya hai.")
```
