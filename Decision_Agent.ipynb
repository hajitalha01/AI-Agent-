```python
class WeatherAgent:
    def decide(self, weather):
        if weather.lower() == "rain":
            return "Umbrella le jao."
        elif weather.lower() == "sunny":
            return "Sunglasses pehen lo."
        else:
            return "Weather clear nahi hai."

agent = WeatherAgent()

weather = input("Weather (rain/sunny): ")
print(agent.decide(weather))
```
