```python
# Cell 1: Imports
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import random

# Set random seeds for reproducibility (optional)
random.seed(42)
np.random.seed(42)
```

```python
# Cell 2: Maze Generator

class MazeGenerator:
    """
    Generates an NxN perfect maze using the Recursive Backtracking (Randomized DFS) algorithm.
    A 'perfect maze' means there are no loops and exactly one unique path between any two cells.
    """
    def __init__(self, size=8):
        self.size = size

    def generate(self):
        # 1 represents a wall, 0 represents an open path
        # We create a grid of all walls initially
        grid = np.ones((self.size, self.size), dtype=int)

        # Keep track of visited cells during DFS
        visited = np.zeros((self.size, self.size), dtype=bool)

        # Stack for backtracking: stores (row, col)
        stack = []

        # Start at position (0, 0)
        start_r, start_c = 0, 0
        grid[start_r, start_c] = 0
        visited[start_r, start_c] = True
        stack.append((start_r, start_c))

        while stack:
            current_r, current_c = stack[-1]

            # Find all unvisited neighbors that are within bounds
            neighbors = []
            # Directions: Up, Down, Left, Right
            directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

            for dr, dc in directions:
                nr, nc = current_r + dr, current_c + dc
                if 0 <= nr < self.size and 0 <= nc < self.size:
                    if not visited[nr, nc]:
                        # To ensure a valid perfect maze layout where paths don't
                        # completely open up into wide spaces, we check if the neighbor
                        # shares too many open neighbors. For a grid-based movement
                        # representation, a standard trick is checking adjacent cells.
                        # Alternatively, standard DFS cuts walls. Here we carve out paths.
                        neighbors.append((nr, nc))

            if neighbors:
                # Choose a random unvisited neighbor
                next_r, next_c = random.choice(neighbors)

                # Check to ensure we don't create loops (keeping it a perfect maze)
                # Count open neighbors of the chosen next cell
                open_neighbors = 0
                for dr, dc in directions:
                    nnr, nnc = next_r + dr, next_c + dc
                    if 0 <= nnr < self.size and 0 <= nnc < self.size:
                        if grid[nnr, nnc] == 0:
                            open_neighbors += 1

                # If it only connects back to our current cell, it's safe to carve
                if open_neighbors <= 1:
                    grid[next_r, next_c] = 0
                    visited[next_r, next_c] = True
                    stack.append((next_r, next_c))
                else:
                    # Mark visited anyway so we don't get stuck in an infinite loop
                    visited[next_r, next_c] = True
            else:
                # No unvisited neighbors, backtrack
                stack.pop()

        # Force start and goal to be open pathways
        grid[0, 0] = 0
        grid[self.size-1, self.size-1] = 0

        return grid
```

```python
# Cell 3: Maze Visualization

def visualize_maze_solution(maze, path):
    """
    Creates an animation of the agent traversing the maze based on a given path.
    """
    fig, ax = plt.subplots(figsize=(6, 6))
    size = maze.shape[0]

    # Create a display grid for visualization
    # 0 = Path (White), 1 = Wall (Black)
    display_grid = np.copy(maze)

    # Render base maze
    im = ax.imshow(display_grid, cmap='binary', origin='upper')

    # Plot Start (Green square) and Goal (Red square)
    ax.plot(0, 0, 'gs', markersize=15, label='Start')
    ax.plot(size-1, size-1, 'rs', markersize=15, label='Goal')

    # Initialize Agent representation (Blue circle)
    agent_marker, = ax.plot([], [], 'bo', markersize=12, label='Agent')

    # Optional path tracking line
    path_line, = ax.plot([], [], 'b--', alpha=0.5)

    ax.set_xticks(range(size))
    ax.set_yticks(range(size))
    ax.grid(True, which='both', color='gray', linestyle='-', linewidth=0.5)
    ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1))

    def init():
        agent_marker.set_data([], [])
        path_line.set_data([], [])
        return agent_marker, path_line

    def update(frame):
        # Frame indexes the step in our path list
        r, c = path[frame]
        agent_marker.set_data([c], [r]) # Matplotlib uses (x, y) mapping to (col, row)

        # Draw trailing history line
        history_r = [pos[0] for pos in path[:frame+1]]
        history_c = [pos[1] for pos in path[:frame+1]]
        path_line.set_data(history_c, history_r)

        ax.set_title(f"Step {frame}/{len(path)-1}")
        return agent_marker, path_line

    # Create the animation
    ani = animation.FuncAnimation(fig, update, frames=len(path), init_func=init, blit=True, interval=300, repeat=False)
    plt.close() # Prevents extra static plot from rendering
    return HTML(ani.to_jshtml())
```

```python
# Cell 4: Environment

class MazeEnvironment:
    """
    Custom grid world environment representing the generated maze layout.
    Tracks state, enforces boundaries, and distributes rewards.
    """
    def __init__(self, maze):
        self.maze = maze
        self.size = maze.shape[0]
        self.start_pos = (0, 0)
        self.goal_pos = (self.size - 1, self.size - 1)
        self.reset()

    def reset(self):
        self.agent_pos = self.start_pos
        self.visited_cells = {self.start_pos: 1} # Tracks visit counts
        self.steps = 0
        return self.agent_pos

    def step(self, action):
        """
        Actions: 0 = Up, 1 = Down, 2 = Left, 3 = Right
        """
        self.steps += 1
        r, c = self.agent_pos

        # Calculate tentative next position
        if action == 0:    # Up
            next_r, next_c = r - 1, c
        elif action == 1:  # Down
            next_r, next_c = r + 1, c
        elif action == 2:  # Left
            next_r, next_c = r, c - 1
        elif action == 3:  # Right
            next_r, next_c = r, c + 1

        # Check boundary collision or wall collision
        if next_r < 0 or next_r >= self.size or next_c < 0 or next_c >= self.size or self.maze[next_r, next_c] == 1:
            # Hit Wall: position remains unchanged
            reward = -10
            done = False
        else:
            # Valid Move
            self.agent_pos = (next_r, next_c)

            if self.agent_pos == self.goal_pos:
                reward = 100
                done = True
            else:
                # Check for repeated cell penalties
                if self.agent_pos in self.visited_cells:
                    reward = -3
                    self.visited_cells[self.agent_pos] += 1
                else:
                    reward = -1
                    self.visited_cells[self.agent_pos] = 1
                done = False

        # Additional step penalty if the agent is wandering endlessly (unnecessary steps)
        if self.steps > (self.size * self.size * 3):
            reward -= 2

        # Hard limit to avoid infinite training episodes
        if self.steps >= 200:
            done = True

        return self.agent_pos, reward, done
```

```python
# Cell 5: Q-Learning Agent

class QLearningAgent:
    """
    Tabular Q-Learning Agent built entirely from scratch.
    """
    def __init__(self, size, alpha=0.1, gamma=0.95, epsilon=1.0, min_epsilon=0.01, decay_rate=0.995):
        self.size = size
        self.alpha = alpha       # Learning rate
        self.gamma = gamma       # Discount factor
        self.epsilon = epsilon   # Exploration rate
        self.min_epsilon = min_epsilon
        self.decay_rate = decay_rate

        # Initialize Q-table: Shape (Size, Size, 4 Actions) initialized to zeros
        self.q_table = np.zeros((size, size, 4))

    def choose_action(self, state):
        """
        Epsilon-Greedy Action Selection
        """
        r, c = state
        if random.random() < self.epsilon:
            # Explore: pick a random action (0 to 3)
            return random.randint(0, 3)
        else:
            # Exploit: pick the action with max Q-value for current state
            # If values are identical, argmax naturally picks the first index
            return np.argmax(self.q_table[r, c])

    def learn(self, state, action, reward, next_state):
        """
        Standard Q-value update formula using Temporal Difference error
        """
        r, c = state
        nr, nc = next_state

        # Max estimated Q-value for the next state
        max_future_q = np.max(self.q_table[nr, nc])

        # Current Q-value
        current_q = self.q_table[r, c, action]

        # Q-learning formula: Q(s,a) = Q(s,a) + alpha * [R + gamma * max(Q(s',a')) - Q(s,a)]
        self.q_table[r, c, action] = current_q + self.alpha * (reward + self.gamma * max_future_q - current_q)

    def decay_epsilon(self):
        """
        Gradually lower exploration rate over time
        """
        if self.epsilon > self.min_epsilon:
            self.epsilon *= self.decay_rate
            self.epsilon = max(self.epsilon, self.min_epsilon)
```

```python
# Cell 6: Training Loop

def train_until_solved(maze, max_episodes=10):
    """
    Trains the agent on the exact same maze structure until it solves it,
    or caps out at the requested number of episodes for inspection.
    """
    env = MazeEnvironment(maze)
    agent = QLearningAgent(size=maze.shape[0])

    best_reward = float('-inf')
    shortest_path = []
    maze_solved = False

    print("="*50)
    print("STARTING TRAINING ON CURRENT MAZE")
    print("="*50)

    episode = 1
    # We continue until the goal is reached at least once AND we hit our run profile metrics
    while episode <= max_episodes or not maze_solved:
        state = env.reset()
        total_reward = 0
        episode_path = [state]
        done = False

        while not done:
            action = agent.choose_action(state)
            next_state, reward, done = env.step(action)

            # Agent updates its core knowledge base
            agent.learn(state, action, reward, next_state)

            state = next_state
            total_reward += reward
            episode_path.append(state)

        # Evaluate goals achieved
        goal_reached = "Yes" if env.agent_pos == env.goal_pos else "No"
        if goal_reached == "Yes":
            maze_solved = True
            if total_reward > best_reward:
                best_reward = total_reward
                shortest_path = episode_path

        # Console output for every individual episode
        print(f"Episode: {episode:<4} | Reward: {total_reward:<5} | Steps: {env.steps:<4} | Goal Reached: {goal_reached:<3} | Epsilon: {agent.epsilon:.3f}")

        # Decay exploration factor
        agent.decay_epsilon()

        # Force progression break condition once requirements are satisfied
        if maze_solved and episode >= max_episodes:
            break

        episode += 1

    print("\n" + "-"*30)
    print("Maze Solved!")
    print(f"Episodes Used: {episode}")
    print(f"Best Reward: {best_reward}")
    print(f"Shortest Path Length: {len(shortest_path)} positions")
    print("-"*30 + "\n")

    return shortest_path
```

```python
# Cell 7: Testing

# Step 1: Create our base Maze Generator engine
generator = MazeGenerator(size=8)

# Step 2: Generate a fresh random perfect 8x8 maze structure
current_maze = generator.generate()

# Step 3: Run the structured training process (Set to 10 episodes as required)
# It will train for at least 10 episodes and won't stop until it finds a valid solution path.
final_path = train_until_solved(current_maze, max_episodes=10)

# Step 4: Display the final animated solution walk down the path discovered
visualize_maze_solution(current_maze, final_path)
```

```python
# Cell 8: Testing with 3 Different Mazes Loop

# Generator initialize karein
generator = MazeGenerator(size=8)
num_mazes_to_solve = 3

for maze_num in range(1, num_mazes_to_solve + 1):
    print("\n" + "="*60)
    print(f" GENERATING AND STARTING MAZE #{maze_num} ")
    print("="*60)

    # 1. New perfect maze generate karein
    current_maze = generator.generate()

    # 2. Agent ko train karein jab tak maze solve na ho jaye (Minimum 10 episodes)
    # Yeh loop tabhi khatam hoga jab goal reach ho chuka ho aur kam se kam 10 episodes pure ho jayein
    final_path = train_until_solved(current_maze, max_episodes=10)

    # 3. Har solve kiye gaye maze ka animation save/display karein
    print(f"Visualizing solution for Maze #{maze_num}...")
    display(visualize_maze_solution(current_maze, final_path))
```
